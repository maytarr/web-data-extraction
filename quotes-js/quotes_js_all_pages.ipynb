```python
import pandas as pd
from bs4 import BeautifulSoup 
from urllib.parse import urljoin #forgot this & forgot to add 'import'
import requests 
import json
```

```python
homepage_url = 'https://quotes.toscrape.com/'
res = requests.get(homepage_url)
print(res.status_code) #wanted to create a variable 'status' instead of printing it, but it's not a biggie.
soup = BeautifulSoup(res.text, 'html.parser')
```

```python
all_quotes = []
base_url = 'https://quotes.toscrape.com/js/page'
current_page_url = 'https://quotes.toscrape.com/js/page/1/' #forgot to add this

while current_page_url:

    print(f'Scraping {current_page_url}')

    res = requests.get(current_page_url)

    soup = BeautifulSoup(res.text, 'html.parser')

    next_button = soup.select_one('li.next a') #forgot to add this

    scripts = soup.select('script')  

    for script in scripts:
        if 'var data' in script.text: #forgot to add this line 
            vardata = script.text
            start = vardata.find('var data =') + len('var data =') #forgot to add the '='
            end = vardata.find('];', start) + 1 #add the +1 to the data_str variable
            data_str = vardata[start:end] #forogt to add vardata 
            quotes = json.loads(data_str)
            all_quotes.extend(quotes)  #wasn't sure what came after passing javascript data to Python object.

    if next_button:
        current_page_url = urljoin(base_url, next_button.get('href')) #added a '+' instead of a ','
    else:
        current_page_url = None 

all_quotes_df = pd.DataFrame(all_quotes)
all_quotes_df.to_csv('allpages_quotes_js.csv', index=False)

```
